```python
import numpy as np
import glob
import mne
```

```python
def create_edf(eeg_data, channel_names, sfreq, output_path):
    """
    Create an EDF file from EEG data.
    
    Parameters:
    - eeg_data: numpy array of shape (n_channels, n_samples)
    - channel_names: list of channel names (length = n_channels)
    - sfreq: sampling frequency in Hz
    - output_path: path to save the EDF file (should end with .edf)
    """
    # Create info structure
    ch_types = ['eeg'] * len(channel_names)
    info = mne.create_info(ch_names=channel_names, sfreq=sfreq, ch_types=ch_types)
    
    # Create RawArray object
    raw = mne.io.RawArray(eeg_data, info)
    
    # Set EEG reference (average reference is common)
    raw.set_eeg_reference('average')
    
    # Save to EDF
    raw.export(output_path, fmt='edf', overwrite=True)
    print(f"EDF file saved to {output_path}")
```
